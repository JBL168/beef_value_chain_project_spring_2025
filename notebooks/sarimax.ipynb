```python
def plot_residual_analysis(actual, predicted, model_name="Model"):
    """
    Create comprehensive residual analysis plots for model evaluation
    
    Parameters:
    actual: array-like, actual values
    predicted: array-like, predicted values
    model_name: str, name of the model for plot titles
    """
    # Calculate residuals
    residuals = actual - predicted
    
    # Create a 2x3 subplot layout
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    fig.suptitle(f'{model_name} - Residual Analysis', fontsize=16, fontweight='bold')
    
    # 1. Residuals vs Fitted Values
    axes[0, 0].scatter(predicted, residuals, alpha=0.6, s=30)
    axes[0, 0].axhline(y=0, color='red', linestyle='--', alpha=0.8)
    axes[0, 0].set_xlabel('Fitted Values')
    axes[0, 0].set_ylabel('Residuals')
    axes[0, 0].set_title('Residuals vs Fitted Values')
    axes[0, 0].grid(True, alpha=0.3)
    
    # Add trend line
    z = np.polyfit(predicted, residuals, 1)
    p = np.poly1d(z)
    axes[0, 0].plot(predicted, p(predicted), "r--", alpha=0.8, linewidth=2)
    
    # 2. Q-Q Plot (Quantile-Quantile)
    stats.probplot(residuals, dist="norm", plot=axes[0, 1])
    axes[0, 1].set_title('Q-Q Plot (Normal Distribution)')
    axes[0, 1].grid(True, alpha=0.3)
    
    # 3. Histogram of Residuals
    axes[0, 2].hist(residuals, bins=30, density=True, alpha=0.7, color='skyblue', edgecolor='black')
    axes[0, 2].set_xlabel('Residuals')
    axes[0, 2].set_ylabel('Density')
    axes[0, 2].set_title('Distribution of Residuals')
    axes[0, 2].grid(True, alpha=0.3)
    
    # Overlay normal distribution
    mu, sigma = np.mean(residuals), np.std(residuals)
    x = np.linspace(residuals.min(), residuals.max(), 100)
    axes[0, 2].plot(x, stats.norm.pdf(x, mu, sigma), 'r-', linewidth=2, label='Normal Distribution')
    axes[0, 2].legend()
    
    # 4. Residuals Over Time
    if hasattr(actual, 'index'):
        time_index = actual.index
    else:
        time_index = range(len(actual))
    
    axes[1, 0].plot(time_index, residuals, marker='o', markersize=3, alpha=0.7)
    axes[1, 0].axhline(y=0, color='red', linestyle='--', alpha=0.8)
    axes[1, 0].set_xlabel('Time')
    axes[1, 0].set_ylabel('Residuals')
    axes[1, 0].set_title('Residuals Over Time')
    axes[1, 0].grid(True, alpha=0.3)
    
    # 5. Actual vs Predicted
    min_val = min(actual.min(), predicted.min())
    max_val = max(actual.max(), predicted.max())
    axes[1, 1].scatter(actual, predicted, alpha=0.6, s=30)
    axes[1, 1].plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=2, label='Perfect Prediction')
    axes[1, 1].set_xlabel('Actual Values')
    axes[1, 1].set_ylabel('Predicted Values')
    axes[1, 1].set_title('Actual vs Predicted')
    axes[1, 1].legend()
    axes[1, 1].grid(True, alpha=0.3)
    
    # 6. Residuals Statistics Box
    axes[1, 2].axis('off')
    
    # Calculate statistics
    mae = mean_absolute_error(actual, predicted)
    mse = mean_squared_error(actual, predicted)
    rmse = np.sqrt(mse)
    mape = mean_absolute_percentage_error(actual, predicted) * 100
    r2 = r2_score(actual, predicted)
    
    # Shapiro-Wilk test for normality
    shapiro_stat, shapiro_p = stats.shapiro(residuals)
    
    # Ljung-Box test for autocorrelation (if time series data)
    try:
        ljung_box = acorr_ljungbox(residuals, lags=10, return_df=True)
        ljung_p = ljung_box['lb_pvalue'].iloc[-1]
    except:
        ljung_p = None
    
    # Create statistics text
    stats_text = f"""
Residual Statistics:
─────────────────────
Mean: {np.mean(residuals):.4f}
Std Dev: {np.std(residuals):.4f}
Min: {np.min(residuals):.4f}
Max: {np.max(residuals):.4f}

Model Performance:
─────────────────────
MAE: {mae:.4f}
RMSE: {rmse:.4f}
MAPE: {mape:.2f}%
R²: {r2:.4f}

Diagnostic Tests:
─────────────────────
Shapiro-Wilk (Normality):
  Statistic: {shapiro_stat:.4f}
  p-value: {shapiro_p:.4f}
  {'✓ Normal' if shapiro_p > 0.05 else '✗ Not Normal'}
"""
    
    if ljung_p is not None:
        stats_text += f"""
Ljung-Box (Independence):
  p-value: {ljung_p:.4f}
  {'✓ Independent' if ljung_p > 0.05 else '✗ Autocorrelated'}
"""
    
    axes[1, 2].text(0.1, 0.5, stats_text, fontsize=10, 
                    verticalalignment='center', fontfamily='monospace',
                    bbox=dict(boxstyle="round,pad=0.3", facecolor="lightgray", alpha=0.8))
    
    plt.tight_layout()
    return fig
```

```python
def plot_autocorrelation_analysis(residuals, model_name="Model", lags=40):
    """
    Plot ACF and PACF of residuals to check for autocorrelation
    """
    fig, axes = plt.subplots(1, 2, figsize=(15, 6))
    fig.suptitle(f'{model_name} - Autocorrelation Analysis of Residuals', fontsize=14, fontweight='bold')
    
    # ACF plot
    plot_acf(residuals, lags=lags, ax=axes[0], alpha=0.05)
    axes[0].set_title('Autocorrelation Function (ACF)')
    axes[0].grid(True, alpha=0.3)
    
    # PACF plot
    plot_pacf(residuals, lags=lags, ax=axes[1], alpha=0.05)
    axes[1].set_title('Partial Autocorrelation Function (PACF)')
    axes[1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    return fig

def plot_forecast_intervals(train_data, test_data, forecast_mean, conf_intervals, model_name="Model"):
    """
    Plot training data, test data, forecasts with confidence intervals
    """
    plt.figure(figsize=(15, 8))
    
    # Plot training data
    plt.plot(train_data.index, train_data, label='Training Data', color='blue', alpha=0.8)
    
    # Plot test data
    plt.plot(test_data.index, test_data, label='Actual Test Data', color='green', linewidth=2)
    
    # Plot forecast
    plt.plot(test_data.index, forecast_mean, label='Forecast', color='red', linestyle='--', linewidth=2)
    
    # Plot confidence intervals
    if conf_intervals is not None:
        plt.fill_between(test_data.index,
                        conf_intervals.iloc[:, 0],
                        conf_intervals.iloc[:, 1],
                        color='red', alpha=0.2,
                        label='95% Confidence Interval')
    
    plt.title(f'{model_name} - Forecast vs Actual', fontsize=14, fontweight='bold')
    plt.xlabel('Date')
    plt.ylabel('Price')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.xticks(rotation=45)
    plt.tight_layout()
    return plt.gcf()

def residual_diagnostics_summary(actual, predicted, model_name="Model"):
    """
    Comprehensive residual diagnostics summary
    """
    residuals = actual - predicted
    
    print(f"\n{'='*60}")
    print(f"RESIDUAL DIAGNOSTICS SUMMARY - {model_name}")
    print(f"{'='*60}")
    
    # Basic statistics
    print(f"\n1. BASIC RESIDUAL STATISTICS:")
    print(f"   Mean: {np.mean(residuals):.6f}")
    print(f"   Std Dev: {np.std(residuals):.4f}")
    print(f"   Min: {np.min(residuals):.4f}")
    print(f"   Max: {np.max(residuals):.4f}")
    print(f"   Skewness: {stats.skew(residuals):.4f}")
    print(f"   Kurtosis: {stats.kurtosis(residuals):.4f}")
    
    # Performance metrics
    print(f"\n2. MODEL PERFORMANCE METRICS:")
    mae = mean_absolute_error(actual, predicted)
    mse = mean_squared_error(actual, predicted)
    rmse = np.sqrt(mse)
    mape = mean_absolute_percentage_error(actual, predicted) * 100
    r2 = r2_score(actual, predicted)
    
    print(f"   MAE: {mae:.4f}")
    print(f"   MSE: {mse:.4f}")
    print(f"   RMSE: {rmse:.4f}")
    print(f"   MAPE: {mape:.2f}%")
    print(f"   R²: {r2:.4f}")
    
    # Normality tests
    print(f"\n3. NORMALITY TESTS:")
    
    # Shapiro-Wilk test
    shapiro_stat, shapiro_p = stats.shapiro(residuals)
    print(f"   Shapiro-Wilk Test:")
    print(f"     Statistic: {shapiro_stat:.4f}")
    print(f"     p-value: {shapiro_p:.6f}")
    print(f"     Result: {'✓ Residuals appear normal' if shapiro_p > 0.05 else '✗ Residuals not normal'}")
    
    # Jarque-Bera test
    jb_stat, jb_p = stats.jarque_bera(residuals)
    print(f"   Jarque-Bera Test:")
    print(f"     Statistic: {jb_stat:.4f}")
    print(f"     p-value: {jb_p:.6f}")
    print(f"     Result: {'✓ Residuals appear normal' if jb_p > 0.05 else '✗ Residuals not normal'}")
    
    # Autocorrelation tests
    print(f"\n4. AUTOCORRELATION TESTS:")
    
    # Ljung-Box test
    try:
        ljung_box = acorr_ljungbox(residuals, lags=10, return_df=True)
        ljung_p = ljung_box['lb_pvalue'].iloc[-1]
        print(f"   Ljung-Box Test (lag 10):")
        print(f"     p-value: {ljung_p:.6f}")
        print(f"     Result: {'✓ No autocorrelation detected' if ljung_p > 0.05 else '✗ Autocorrelation present'}")
    except Exception as e:
        print(f"   Ljung-Box Test: Error - {e}")
    
    # Durbin-Watson test
    try:
        from statsmodels.stats.stattools import durbin_watson
        dw_stat = durbin_watson(residuals)
        print(f"   Durbin-Watson Test:")
        print(f"     Statistic: {dw_stat:.4f}")
        if 1.5 <= dw_stat <= 2.5:
            print(f"     Result: ✓ No strong autocorrelation")
        else:
            print(f"     Result: ✗ Possible autocorrelation")
    except Exception as e:
        print(f"   Durbin-Watson Test: Error - {e}")
    
    print(f"\n{'='*60}\n")
```

```python
file_path = os.path.join('..', 'aggregated_data.csv')
data = pd.read_csv(file_path)

# inflation adjustment
base_cpi = data['CPI_US'].iloc[-1]
data['AVG_PRICE_AUCTION'] = data['AVG_PRICE_AUCTION'] * (base_cpi/data['CPI_US'])
data['DATE'] = pd.to_datetime(data['DATE'])

data = data.sort_values('DATE', ignore_index=True).set_index('DATE').drop( # drop columns not used for training
    ['avg_price_steers_ML1', 
     'avg_price_steers_ML1_2', 
     'avg_price_heifers_ML1', 
     'avg_price_heifers_ML1_2', 
     'NATIONAL_AUCTION_SALES', 
     'NATIONAL_TOTAL_SALES', 
     'TEXAS_AUCTION_SALES', 
    #  'CORN_FUTURES_TYPICAL_PRICE', 
    #  'DIESEL_RETAIL_PRICE', 
     'ALFAFA_HAY_PPI', 
     'CORN_PPI',
    #  'FEEDER_CATTLE_FUTURES_TYPICAL_PRICE',
     'US_MXN_RATES',
     'CPI_US',
     'FEDERAL_INTEREST_RATE',
     'US_MXN_RATES',
     ], axis=1)
```

```python
def evaluate_forecast(model_name, actual, predicted):
    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    r2 = r2_score(actual,predicted)
    
    acc = 100.0 * (1 - mae / actual.mean()) if actual.mean() != 0 else float('nan')
    print(f"\n[{model_name}] Test Performance:")
    print(f"MAE: {mae:.2f}, RMSE: {rmse:.2f}, R²: {r2:.2f}, Accuracy: {acc:.2f}%")
```

```python
# FIND WHICH DATA POINTS ARE MISSING
start_date = data.index.min()
end_date = data.index.max()
full_index = pd.date_range(start=start_date, end=end_date, freq='W-FRI')
missing_dates = full_index.difference(data.index)

# REINDEX DATA & IMPUTE MISSING VALUES

arima_data = data.reindex(full_index).interpolate(method='linear')

# freq = pd.infer_freq(arima_data.index)

```

```python
n = len(arima_data)
train_size = int(n * 0.9)

train = arima_data.iloc[:train_size]
test = arima_data.iloc[train_size:]

train['AVG_PRICE_AUCTION'].shape
```

```python
model = auto_arima(y=train['AVG_PRICE_AUCTION'],
                  start_p=0, start_q=0,
                  max_p=5, max_q=5,     # Reduced from 5 to limit complexity
                  m=12,                 
                  seasonal=True,
                  max_P=2, max_Q=2,     # Add limits to seasonal components
                  d=None,                  # Force first differencing
                  D=None,                  # Force seasonal differencing
                  trace=True,
                  error_action='ignore',
                  suppress_warnings=True,
                  stepwise=True,
                  random_state=42,
                  information_criterion='aic',  # Use AIC for model selection
                  n_fits=30)            # Reduced number of fits
```

```python
# CREATE AND TRAIN ARIMA MODEL
exog_features = ['CORN_FUTURES_TYPICAL_PRICE', 'DIESEL_RETAIL_PRICE', 'FEEDER_CATTLE_FUTURES_TYPICAL_PRICE']

model = ARIMA(train['AVG_PRICE_AUCTION'], 
              exog=train[exog_features],
              order=(0, 1, 1), 
              seasonal_order=(1, 0, 0, 12))

model_fit = model.fit()
# print(model_fit.summary())

# FORECAST DATA
forecast_results = model_fit.get_forecast(steps=len(test),
                                          exog=test[exog_features])
mean_forecast = forecast_results.predicted_mean
confidence_intervals = forecast_results.conf_int(alpha=0.05)  # 95% confidence interval
```
